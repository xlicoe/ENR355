```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# PID simulation and plotting
def run_pid_simulation(Kp, Ki, Kd):
    # ==========================================
    # 1. Simulation & System Parameters
    # ==========================================
    dt = 0.1
    t = np.arange(0, 50, dt)
    n = len(t)

    mass = 1.0
    friction = 0.2

    # ==========================================
    # 2. Setup Data Arrays
    # ==========================================
    setpoint = np.zeros(n)
    setpoint[int(10/dt):] = 10.0 # Step change at t=10s

    position = np.zeros(n)
    velocity = np.zeros(n)

    integral_error = 0
    prev_error = 0

    # ==========================================
    # 3. Simulation Loop
    # ==========================================
    for i in range(1, n):
        # Calculate error
        error = setpoint[i] - position[i-1]

        # PID Terms
        P = Kp * error
        integral_error += error * dt
        I = Ki * integral_error
        D = Kd * (error - prev_error) / dt

        # Compute control variable (u) with anti-windup limits
        u = np.clip(P + I + D, -15, 15)

        # Process / Plant Simulation (F = ma)
        acceleration = (u - friction * velocity[i-1]) / mass
        velocity[i] = velocity[i-1] + acceleration * dt
        position[i] = position[i-1] + velocity[i] * dt

        # Save error for next loop
        prev_error = error

    # ==========================================
    # 4. Plotting the Results
    # ==========================================
    # Adjusted figure size for a single, wider plot
    plt.figure(figsize=(10, 5))

    plt.plot(t, setpoint, 'r--', linewidth=2, label='Setpoint (Target)')
    plt.plot(t, position, 'b-', linewidth=2, label='Actual Position')
    plt.title(f'Interactive PID Control Response', fontsize=14)
    plt.xlabel('Time (seconds)')
    plt.ylabel('Position')



    plt.legend(loc='lower right')
    plt.grid(True)

    plt.tight_layout()
    plt.show()

# ==========================================
# 5. Create Interactive Sliders
# ==========================================
interact(run_pid_simulation,
         Kp=FloatSlider(value=2.0, min=0.0, max=10.0, step=0.1, description='Kp (Prop):', continuous_update=False),
         Ki=FloatSlider(value=0.4, min=0.0, max=5.0, step=0.05, description='Ki (Integ):', continuous_update=False),
         Kd=FloatSlider(value=1.5, min=0.0, max=10.0, step=0.1, description='Kd (Deriv):', continuous_update=False));
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def run_pid_simulation(Kp, Ki, Kd):
    # ==========================================
    # 1. Simulation & System Parameters
    # ==========================================
    dt = 0.1
    t = np.arange(0, 50, dt)
    n = len(t)

    mass = 1.0
    friction = 0.2

    # ==========================================
    # 2. Setup Data Arrays
    # ==========================================
    setpoint = np.zeros(n)
    setpoint[int(10/dt):] = 10.0 # Step change at t=10s

    position = np.zeros(n)
    velocity = np.zeros(n)

    integral_error = 0
    prev_error = 0

    # ==========================================
    # 3. Simulation Loop
    # ==========================================
    for i in range(1, n):
        # Calculate error
        error = setpoint[i] - position[i-1]

        # PID Terms
        P = Kp * error
        integral_error += error * dt
        I = Ki * integral_error
        D = Kd * (error - prev_error) / dt

        # Compute control variable (u) with anti-windup limits
        u = np.clip(P + I + D, -15, 15)

        # --- NEW: Introduce a Disturbance ---
        # A sudden "wind gust" pushes the object between t=30 and t=32 seconds
        disturbance = 0
        if 30.0 <= t[i] <= 32.0:
            disturbance = -5.0 # Pushing against the direction of travel

        # Process / Plant Simulation (F = ma)
        # Total force now includes the disturbance
        acceleration = (u + disturbance - friction * velocity[i-1]) / mass
        velocity[i] = velocity[i-1] + acceleration * dt
        position[i] = position[i-1] + velocity[i] * dt

        # Save error for next loop
        prev_error = error

    # ==========================================
    # 4. Plotting the Results
    # ==========================================
    plt.figure(figsize=(10, 5))

    # Highlight the disturbance on the plot
    plt.axvspan(30, 32, color='gray', alpha=0.2, label='Wind Gust Disturbance')

    plt.plot(t, setpoint, 'r--', linewidth=2, label='Setpoint (Target)')
    plt.plot(t, position, 'b-', linewidth=2, label='Actual Position')
    plt.title('Interactive PID Control Response (with Disturbance)', fontsize=14)
    plt.xlabel('Time (seconds)')
    plt.ylabel('Position')

    # Y-axis will auto-scale to show the whole response

    plt.legend(loc='lower right')
    plt.grid(True)

    plt.tight_layout()
    plt.show()

# ==========================================
# 5. Create Interactive Sliders
# ==========================================
interact(run_pid_simulation,
         Kp=FloatSlider(value=2.0, min=0.0, max=10.0, step=0.1, description='Kp (Prop):', continuous_update=False),
         Ki=FloatSlider(value=0.4, min=0.0, max=5.0, step=0.05, description='Ki (Integ):', continuous_update=False),
         Kd=FloatSlider(value=1.5, min=0.0, max=10.0, step=0.1, description='Kd (Deriv):', continuous_update=False));
```
